```python
#for döngüsü ile range kullanarak dizi kullanmadan klavyeden girilen n adet sayunun ortalamasını bulunuz.(n deperi de klavyeden girilsin)
n = int(input("Kaç tane sayının ortalaması alınacak: "))
toplam = 0
for i in range(n):
  sayi = float(input(f"{i+1}. sayıyı giriniz: "))
  toplam += sayi
print("Ortalama:" , {toplam/n})
```

## **While döngüsü**

```python
sayac = 0
while sayac < 5 :
  print("Sayaç:" , sayac)
  sayac +=1
```

```python
#kullanıcı 0 girene kadar sayıları toopla

toplam = 0
while True:
  sayi = int(input("Sayı giriniz (0 çıkış): "))
  if sayi == 0:
    break
  toplam += sayi
print("Toplam:", toplam)
```

```python
#1 den 10 a kadar sayıları yazdır 5 i gördüğünde dur
for i in range (1,11):
    if i == 5:
      break
    print(i)
```

```python
#while döngüsü ile range kullanarak dizi kullanmadan klavyeden girilen n adet sayının ortalamasını bulunuz(n klavyeden girilsin)
n = int(input("Kaç tane sayının ortalaması alınacak: "))
toplam = 0
i = 0
while i < n:
  sayi = float(input(f"{i+1}Sayı giriniz: "))
  toplam += sayi
  i +=1
  ortalama = toplam/n
print("Ortalama:" , ortalama)
```
